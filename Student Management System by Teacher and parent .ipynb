```python
"""
Student Management System - By Teacher and Parents
-----------------------------------------------------
This system supports two types of users:

1. TEACHER (Admin access):
   - Add student details
   - Update student records
   - Delete student records
   - Search for a student by ID
   - Display all students

2. PARENT (Limited access):
   - Search for their child's record by Student ID
   - View their child's details only (read-only access)

Data is stored persistently in a CSV file (students.csv).
"""

import csv
import os

FILENAME = "students.csv"
FIELDNAMES = ["ID", "Name", "Age", "Course", "Email", "ParentContact"]


def init_file():
    """Create the CSV file with headers if it does not already exist."""
    if not os.path.exists(FILENAME):
        with open(FILENAME, mode="w", newline="") as file:
            writer = csv.DictWriter(file, fieldnames=FIELDNAMES)
            writer.writeheader()


def load_students():
    """Load and return the list of all students from the CSV file."""
    init_file()
    with open(FILENAME, mode="r", newline="") as file:
        reader = csv.DictReader(file)
        return list(reader)


def save_students(students):
    """Write the complete list of students back to the CSV file (overwrite)."""
    with open(FILENAME, mode="w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=FIELDNAMES)
        writer.writeheader()
        writer.writerows(students)


def print_student_details(s):
    """Print a single student's details in a readable format."""
    print("Student Details:")
    print(f"  ID             : {s['ID']}")
    print(f"  Name           : {s['Name']}")
    print(f"  Age            : {s['Age']}")
    print(f"  Course         : {s['Course']}")
    print(f"  Email          : {s['Email']}")
    print(f"  Parent Contact : {s['ParentContact']}\n")


# ---------------------------------------------------------
# TEACHER FUNCTIONS (Full access: Add, Update, Delete, etc.)
# ---------------------------------------------------------

def add_student():
    """Add a new student record after checking for duplicate IDs."""
    students = load_students()

    student_id = input("Enter Student ID: ").strip()

    # Check if the ID already exists
    for s in students:
        if s["ID"] == student_id:
            print(f"Error: Student ID '{student_id}' already exists!\n")
            return

    name = input("Enter Name: ").strip()
    age = input("Enter Age: ").strip()
    course = input("Enter Course: ").strip()
    email = input("Enter Email: ").strip()
    parent_contact = input("Enter Parent Contact Number: ").strip()

    new_student = {
        "ID": student_id,
        "Name": name,
        "Age": age,
        "Course": course,
        "Email": email,
        "ParentContact": parent_contact,
    }

    students.append(new_student)
    save_students(students)
    print(f"Success: Student '{name}' added successfully!\n")


def update_student():
    """Update an existing student's record by ID."""
    students = load_students()
    student_id = input("Enter Student ID to update: ").strip()

    for s in students:
        if s["ID"] == student_id:
            print(f"Current details: {s}")
            print("(Enter a new value, or press Enter to keep the current value)")

            name = input(f"New Name [{s['Name']}]: ").strip()
            age = input(f"New Age [{s['Age']}]: ").strip()
            course = input(f"New Course [{s['Course']}]: ").strip()
            email = input(f"New Email [{s['Email']}]: ").strip()
            parent_contact = input(f"New Parent Contact [{s['ParentContact']}]: ").strip()

            if name:
                s["Name"] = name
            if age:
                s["Age"] = age
            if course:
                s["Course"] = course
            if email:
                s["Email"] = email
            if parent_contact:
                s["ParentContact"] = parent_contact

            save_students(students)
            print(f"Success: Student ID '{student_id}' updated successfully!\n")
            return

    print(f"Error: Student ID '{student_id}' not found!\n")


def delete_student():
    """Delete a student record by ID."""
    students = load_students()
    student_id = input("Enter Student ID to delete: ").strip()

    updated_students = [s for s in students if s["ID"] != student_id]

    if len(updated_students) == len(students):
        print(f"Error: Student ID '{student_id}' not found!\n")
        return

    save_students(updated_students)
    print(f"Success: Student ID '{student_id}' deleted successfully!\n")


def search_student_teacher():
    """Search for a student by ID and display full details (Teacher access)."""
    students = load_students()
    student_id = input("Enter Student ID to search: ").strip()

    for s in students:
        if s["ID"] == student_id:
            print_student_details(s)
            return

    print(f"Error: Student ID '{student_id}' not found!\n")


def display_all_students():
    """Display all student records in a formatted table (Teacher access)."""
    students = load_students()

    if not students:
        print("No student records found.\n")
        return

    print("\n" + "-" * 90)
    print(f"{'ID':<10}{'Name':<18}{'Age':<6}{'Course':<15}{'Email':<22}{'Parent Contact':<15}")
    print("-" * 90)
    for s in students:
        print(f"{s['ID']:<10}{s['Name']:<18}{s['Age']:<6}{s['Course']:<15}{s['Email']:<22}{s['ParentContact']:<15}")
    print("-" * 90 + "\n")


def teacher_menu():
    """Menu for Teacher role with full access to student records."""
    while True:
        print("\n========== TEACHER MENU ==========")
        print("1. Add Student")
        print("2. Update Student")
        print("3. Delete Student")
        print("4. Search Student by ID")
        print("5. Display All Students")
        print("6. Back to Main Menu")
        print("====================================")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            update_student()
        elif choice == "3":
            delete_student()
        elif choice == "4":
            search_student_teacher()
        elif choice == "5":
            display_all_students()
        elif choice == "6":
            print("Returning to Main Menu...\n")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 6.\n")


# ---------------------------------------------------------
# PARENT FUNCTIONS (Limited access: View child's record only)
# ---------------------------------------------------------

def search_student_parent():
    """Allow a parent to search and view only their child's record."""
    students = load_students()
    student_id = input("Enter your child's Student ID: ").strip()

    for s in students:
        if s["ID"] == student_id:
            print_student_details(s)
            return

    print(f"Error: No student record found for ID '{student_id}'!\n")


def parent_menu():
    """Menu for Parent role with read-only access to their child's record."""
    while True:
        print("\n========== PARENT MENU ==========")
        print("1. Search Child's Record by Student ID")
        print("2. Back to Main Menu")
        print("===================================")

        choice = input("Enter your choice (1-2): ").strip()

        if choice == "1":
            search_student_parent()
        elif choice == "2":
            print("Returning to Main Menu...\n")
            break
        else:
            print("Invalid choice! Please enter 1 or 2.\n")


# ---------------------------------------------------------
# MAIN MENU (Role Selection)
# ---------------------------------------------------------

def main_menu():
    """Display the main menu and route the user to Teacher or Parent menu."""
    init_file()
    while True:
        print("========== STUDENT MANAGEMENT SYSTEM ==========")
        print("Login as:")
        print("1. Teacher")
        print("2. Parent")
        print("3. Exit")
        print("=================================================")

        choice = input("Enter your choice (1-3): ").strip()

        if choice == "1":
            teacher_menu()
        elif choice == "2":
            parent_menu()
        elif choice == "3":
            print("Exiting program... Data has been saved in students.csv")
            break
        else:
            print("Invalid choice! Please enter a number between 1 and 3.\n")


# Running this cell in Jupyter Notebook will start the menu
if __name__ == "__main__":
    main_menu()
```
